```python
#randomly

print("Welcome to the PyPasswoard Generator !")
letters = int(input("How many letters would you like in your password?\n"))
symbols = int(input("How many symbols would you like?\n"))
numbers = int(input("How many numbers would you like?\n"))

password_list = []

for char in range(1, letters + 1):
    password_list.append(random.choice("abcdefghijklmnopqrstuvwxyz"))

for char in range(1, symbols + 1):
    password_list.append(random.choice("!@#$%^&*()"))

for char in range(1, numbers + 1):
    password_list.append(random.choice("1234567890"))

print(password_list)
random.shuffle(password_list) #shuffles the list
print(password_list)

#back to string
password = ""
for char in password_list:
    password += char

print(f"Your password is: {password}")

```
